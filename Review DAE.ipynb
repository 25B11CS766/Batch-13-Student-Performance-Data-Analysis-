```python
#IMPORTING LIBRARIES AND LOADING DATA

import pandas as pd
import matplotlib.pyplot as plt

# Load Excel dataset
file_path = "/content/Student performance data.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

# Display first 5 records
print(df.head())
```

```python
#CLEANING DATA

# Check dataset information
print("\nDataset Information:")
print(df.info())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

print("\nData cleaning completed!")
print("Records after cleaning:", len(df))
```

```python
#SCORE ANALYSIS

print("\n----- SCORE ANALYSIS -----")

print("Highest Final Score:", df['final_score'].max())
print("Lowest Final Score:", df['final_score'].min())
print("Average Final Score:", round(df['final_score'].mean(), 2))
print("Median Final Score:", df['final_score'].median())

print("\nFinal Score Statistics:")
print(df['final_score'].describe())
```

```python
#AVERAGE FINAL SCORE

average_final_score = df['final_score'].mean()

print("\nAverage Final Score:",
      round(average_final_score, 2))
```

```python
#NUMBER OF STUDENTS PASSED

passed_students = (df['passed'] == 'Yes').sum()

print("Number of Students Passed:",
      passed_students)
```

```python
#NUMBER OF STUDENTS FAILED

failed_students = (df['passed'] == 'No').sum()

print("Number of Students Failed:",
      failed_students)
```

```python
#PASS RATE

total_students = len(df)

pass_rate = (passed_students / total_students) * 100

print("Pass Rate:",
      round(pass_rate, 2), "%")
```

```python
# Pass / Fail Distribution — Doughnut Chart

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("/content/Student performance data.xlsx")

# Count passed and failed students
passed = (
    df["passed"].astype(str).str.strip().str.lower() == "yes"
).sum()

failed = (
    df["passed"].astype(str).str.strip().str.lower() == "no"
).sum()

# Create doughnut chart
labels = ["Passed", "Failed"]
values = [passed, failed]
colors = ["#20C568", "#F04444"]

plt.figure(figsize=(7, 6))

plt.pie(
    values,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.45}
)

plt.title("Pass / Fail Distribution")
plt.tight_layout()
plt.show()
```

```python
# Average Score by Gender — Bar Chart

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("/content/Student performance data.xlsx")

# Calculate average score by gender
gender_scores = df.groupby("gender")["final_score"].mean()

# Create bar chart
plt.figure(figsize=(7, 5))

plt.bar(
    gender_scores.index,
    gender_scores.values,
    color="#6666F5"
)

plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Score")

plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Attendance vs Final Score — Scatter Plot

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("/content/Student performance data.xlsx")

# Remove missing values
data = df.dropna(subset=["attendance_rate", "final_score"])

# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    data["attendance_rate"],
    data["final_score"],
    color="#9055FF",
    alpha=0.7,
    s=25
)

plt.title("Attendance vs Final Score")
plt.xlabel("Attendance Rate (%)")
plt.ylabel("Final Score")

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Previous Score vs Final Score — Scatter Plot

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("/content/Student performance data.xlsx")

# Remove missing values
data = df.dropna(subset=["previous_score", "final_score"])

# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    data["previous_score"],
    data["final_score"],
    color="#FF9D00",
    alpha=0.7,
    s=25
)

plt.title("Previous Score vs Final Score")
plt.xlabel("Previous Score")
plt.ylabel("Final Score")

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
